# Larger_Scatter_Density — analysis
Density proxy: $\rho \propto T / \phi_{\max}^2$.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from pathlib import Path
HERE = Path('.').resolve()
ev = pd.read_csv(HERE/'results'/'lsd_events.csv')
ev['rho'] = ev['n_trk'] / ev['phi_max']**2
print(ev.shape); ev.head()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
for (phi, ss), sub in ev.groupby(['phi_max','sigma_scatt']):
    g = sub.groupby('n_trk').agg(
        effC=('cls_default_segment_efficiency','mean'),
        effQ=('q_default_segment_efficiency','mean')).reset_index()
    lab = f"phi={phi:g} ss={ss:g}"
    axes[0].plot(g['n_trk'], g['effC'], 'o-', label=lab)
    axes[1].plot(g['n_trk'], g['effQ'], 's-', label=lab)
for ax, t in zip(axes, ['classical seg eff', 'quantum seg eff']):
    ax.set_xscale('log'); ax.set_xlabel('T'); ax.set_title(t)
    ax.legend(fontsize=7); ax.grid(alpha=.3)
plt.tight_layout(); plt.show()
```

```python
# Efficiency vs density
fig, ax = plt.subplots(figsize=(7,4))
for ss, sub in ev.groupby('sigma_scatt'):
    g = sub.groupby(['phi_max','n_trk']).agg(
        eff=('q_default_segment_efficiency','mean'),
        rho=('rho','first')).reset_index()
    ax.scatter(g['rho'], g['eff'], label=f'ss={ss:g}', alpha=.7)
ax.set_xscale('log'); ax.set_xlabel(r'$\rho = T/\phi_{\max}^2$')
ax.set_ylabel('quantum seg eff'); ax.legend(); ax.grid(alpha=.3)
plt.tight_layout(); plt.show()
```
